```python
def count_subsets(n, arr, val):
    count = 0
    for i in range(2**n):
        s = 0
        for j in range(n):
            if i & (2**j):
                s = s + arr[j]
        if s == val:
            count += 1
    return count

t = int(input())
for _ in range(t):
    n = int(input())
    arr = list(map(int, input().split()))
    val = int(input())
    result = count_subsets(n,arr, val)
    if result == 0:
        print('No subset is found')
    else:
        print(result)


# not clearly understand still now
```

```python
def count(pos,last,n,r,end):
    if (pos==n):
        if(last==end):
            return(1)
        else:
            return(0)
    total=0
    for i in range(1,r+1):
        if(i!=last):
            total+=count(pos+1,i,n,r,end)
    return(total)


n=int(input())
r=int(input())
end=int(input())
result=count(1,1,n,r,end)
print(result)
```
